# Preprocessing — QA Chatbot
This notebook demonstrates **Step 1–3** of the retrieval pipeline: loading documents, cleaning text and tokenisation. Run top to bottom.

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
from src.preprocessing import load_documents, clean_text, tokenize, split_sentences
from src.utils import DOC_DIR
```

## Step 1: Load documents
Each `.txt` file is split into paragraphs; every paragraph becomes one retrievable *passage*.

```python
passages = load_documents(DOC_DIR)
print('Total passages:', len(passages))
passages[0]
```

## Step 2: Clean text
Lower-case, strip punctuation, collapse whitespace.

```python
sample = passages[1]['text']
print('RAW:\n', sample)
print('\nCLEANED:\n', clean_text(sample))
```

## Step 3: Tokenisation
Stop-word removal + light stemming.

```python
print(tokenize(sample))
```

## Sentence splitting
Used later by the retriever to extract a single-sentence answer from a passage.

```python
print(split_sentences(sample))
```
